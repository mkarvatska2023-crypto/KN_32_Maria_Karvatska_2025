- робимо прототип гри (карткової гри) де у нас є ітеми (наприклад якась зброя) яка має певні властивості і якою ми можемо робити ходи і загатись (цими ітемами)

```python
import random
names = ["Alice", "Bob", "Charlie", "Diana", "Eve"]
magic_attributes = ["fire", "ice", "lightning", "earth", "wind"]



def draw_card():
    item = {
    "name":random.choice(names),
    "attack_power":random.randint(5, 15),
    "durability":random.randint(10, 100),
    "magic_atribute": f"Magic: {random.choice(names)}"
}
print(f"""
      Витяуємо карту:
      Нам випала {item['name']} 
      з атакою {item['attack_power']}, 
      міцністю {item['durability']} 
      та атрибутом {item['magic_atribute']}
      """)

return item 

```
