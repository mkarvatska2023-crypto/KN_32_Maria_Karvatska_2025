```python
class User:
    def __init__(self, name):
        self.name = name
        self.is_logged = False

    def login(self):
        self.is_logged = True
        print(f"{self.name} увійшов у систему")

    def logout(self):
        self.is_logged = False
        print(f"{self.name} вийшов із системи")

    @property
    def info(self):
        return f"Користувач: {self.name}, статус: {'online' if self.is_logged else 'offline'}"


class Admin(User):
    def __init__(self, name, level):
        super().__init__(name)
        self.level = level

    def ban_user(self, user_name):
        print(f"Адмін {self.name} заблокував користувача {user_name}")

    def reset_system(self):
        print(f"Адмін {self.name} виконав скидання системи")

    @property
    def info(self):
        return f"{super().info}, рівень доступу: {self.level}"


class Moderator(User):
    def delete_comment(self, comment_id):
        print(f"Модератор {self.name} видалив коментар #{comment_id}")

    def warn_user(self, user_name):
        print(f"Модератор {self.name} попередив користувача {user_name}")


# створення об'єктів
u = User("Оля")
a = Admin("Богдан", 5)
m = Moderator("Іван")

# перевірка роботи
print(u.info)
u.login()

print(a.info)
a.login()
a.ban_user("Марія")
a.reset_system()

print(m.info)
m.login()
m.delete_comment(10)
m.warn_user("Петро")
```

# Поліморфізм

- Напишемо програму, яка імітує систему користувачів у застосунку, їх авторизацію та взаємодію з різними функціями системи.

- Ми створимо базовий клас користувача та розширимо його різними типами користувачів (адміністратор, модератор, звичайний користувач), які виконують різні дії в системі.

- Першим видом поліморфізму є перевантаження методів через різну кількість аргументів, наприклад при реєстрації або авторизації користувача з різними параметрами.

- Другим видом є поліморфізм через наслідування, коли різні типи користувачів мають однакові методи (наприклад login() або get_info()), але різну реалізацію: адміністратор має додаткові права, а модератор виконує контроль контенту.

- Третім видом є Duck Typing, коли різні об’єкти (користувач, адміністратор, гість) можуть використовуватися взаємозамінно, якщо вони реалізують однакові методи, наприклад login() або logout().

```python
class User:
    def __init__(self, name):
        self.name = name

    def action(self, target):
        print(f"{self.name} виконує дію над {target.name}")


class Admin:
    def __init__(self, name):
        self.name = name

    def action(self, target):
        print(f"Адмін {self.name} блокує користувача {target.name}")


class Moderator:
    def __init__(self, name):
        self.name = name

    def action(self, target):
        print(f"Модератор {self.name} перевіряє користувача {target.name}")


class BrokenUser:
    def __init__(self, name):
        self.name = name
        # немає нормальної реалізації action


class System:
    def execute_action(self, user, target):
        print(f"Система запускає дію від {user.name}")
        user.action(target)


# Використання (Duck Typing)
u = User("Оля")
a = Admin("Богдан")
m = Moderator("Іван")

target = User("Марія")

s = System()

s.execute_action(u, target)
s.execute_action(a, target)
s.execute_action(m, target)
```

```python
class UserSystem:
    def __init__(self, name):
        self.name = name
        self.__base_permissions = 1

    def perform_action(self, target):
        print(f"{self.name} виконує базову дію над {target.name}")
        target.react(self.__base_permissions)


class AdminUser(UserSystem):
    def __init__(self, name, extra_rights):
        super().__init__(name)
        self.extra_rights = extra_rights

    # перевизначення методу (поліморфізм)
    def perform_action(self, target):
        print(f"{self.name} (адмін) виконує розширену дію над {target.name}")
        target.react(self.extra_rights + 2)


class ModeratorUser(UserSystem):
    def perform_action(self, target):
        print(f"{self.name} (модератор) перевіряє {target.name}")
        target.react(2)


class TargetUser:
    def __init__(self, name):
        self.name = name

    def react(self, power):
        print(f"{self.name} отримав вплив рівня {power}")


# Використання
u = UserSystem("Оля")
a = AdminUser("Богдан", 5)

t = TargetUser("Марія")

u.perform_action(t)
a.perform_action(t)
```

# Абстракція

- Спочатку створимо абстрактний клас BaseLLMModel, який описує загальну поведінку будь-якої LLM моделі.
Він містить абстрактні методи ask_llm() та generate_response().

- Потім реалізуємо кілька конкретних моделей, які наслідують цей клас.

```python
from abc import ABC, abstractmethod

class BaseLLMModel(ABC):

    @abstractmethod
    def ask_llm(self, prompt: str) -> str:
        pass

    @abstractmethod
    def generate_response(self, prompt: str) -> str:
        pass


class GPTModel(BaseLLMModel):
    def ask_llm(self, prompt: str) -> str:
        return f"GPT отримав запит: {prompt}"

    def generate_response(self, prompt: str) -> str:
        return f"GPT відповідь: {prompt[::-1]}"


model = GPTModel()
print(model.ask_llm("Привіт"))
print(model.generate_response("Hello"))
```

```python
from typing import Protocol
from abc import ABC, abstractmethod


class LLMModelProtocol(Protocol):
    @property
    def model_name(self) -> str:
        ...

    def tokenize(self, text: str) -> list:
        ...


class ModelInterface(ABC):
    @abstractmethod
    def askLLm(self, question: str) -> str:
        pass

    @abstractmethod
    def generateResponse(self, prompt: str) -> str:
        pass


class GPT3Model:
    def __init__(self, model_name):
        self._model_name = model_name

    @property
    def model_name(self) -> str:
        return self._model_name

    def tokenize(self, text: str) -> list:
        return text.lower().split()


class BERTModel:
    def __init__(self, model_name):
        self._model_name = model_name

    @property
    def model_name(self) -> str:
        return self._model_name

    def tokenize(self, text: str) -> list:
        return [word.strip("?,.!") for word in text.split()]


# 🔹 Рівень обробки запиту (додаткова абстракція логіки)
class QueryProcessor:
    def preprocess(self, text: str) -> str:
        return text.strip()


# 🔹 Чат-інтерфейс (працює з будь-якою LLM через Protocol)
class ChatInterface:
    def __init__(self, llm_model: LLMModelProtocol):
        self.llm_model = llm_model
        self.processor = QueryProcessor()

    def askLLm(self, question: str) -> str:
        clean_question = self.processor.preprocess(question)
        tokens = self.llm_model.tokenize(clean_question)
        return f"{self.llm_model.model_name} відповідає: {tokens}"

    def generateResponse(self, prompt: str) -> str:
        clean_prompt = self.processor.preprocess(prompt)
        tokens = self.llm_model.tokenize(clean_prompt)
        return f"{self.llm_model.model_name} генерує відповідь: {tokens}"


# 🔹 Поліморфне використання моделей
for model in [GPT3Model("GPT-3"), BERTModel("BERT")]:
    chat = ChatInterface(model)

    print(chat.askLLm("  Як працює LLM? "))
    print(chat.generateResponse("Які є переваги LLM?"))
```

```python
from typing import Protocol
import random


class LLMModelProtocol(Protocol):
    @property
    def model_name(self) -> str:
        ...

    def generate_response(self, text: str) -> str:
        ...


class SimpleLLM:
    def __init__(self, model_name):
        self._model_name = model_name  # інкапсуляція

    @property
    def model_name(self) -> str:
        return self._model_name

    def generate_response(self, text: str) -> str:
        return f"{self._model_name} відповідає на: {text}"


# 🔹 інтерфейс роботи з LLM
class ChatSystem:
    def __init__(self, model: LLMModelProtocol):
        self.model = model

    def send_request(self, text: str):
        return self.model.generate_response(text)


# створення моделі
llm = SimpleLLM("GPT-mini")
chat = ChatSystem(llm)


# 🔹 генератор випадкових запитів
questions = [
    "Що таке Python?",
    "Як працює AI?",
    "Поясни ООП",
    "Що таке LLM?",
    "Навіщо потрібні моделі?"
]

# 🔹 цикл з випадковими запитами
for _ in range(5):
    question = random.choice(questions)
    print(chat.send_request(question))
```

- створено клас SimpleLLM, який зберігає назву моделі як приватний атрибут. Доступ до нього здійснюється через властивість model_name, що демонструє інкапсуляцію.

```python
import random

class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance  # інкапсуляція

    def deposit(self, amount):
        self.__balance += amount
        print(f"Поповнення: +{amount}")

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
            print(f"Зняття: -{amount}")
        else:
            print(f"Спроба зняття {amount} — недостатньо коштів")

    def get_balance(self):
        return self.__balance


# створення рахунку
account = BankAccount("Bohdan", 1000)

# 10 випадкових операцій
for _ in range(10):
    operation = random.choice(["deposit", "withdraw"])
    amount = random.randint(50, 500)

    if operation == "deposit":
        account.deposit(amount)
    else:
        account.withdraw(amount)

# кінцевий результат
print("\nКінцевий баланс:", account.get_balance())
```

1. # Використання парадигм для створення простої гри

```python
from abc import ABC, abstractmethod
from random import randint


# 🔹 Абстрактний клас персонажа
class Character(ABC):
    def __init__(self, name, health=500):
        self.name = name
        self.health = health

    @abstractmethod
    def attack(self, enemy):
        pass


# 🔹 Наслідування + інкапсуляція
class Knight(Character):
    def __init__(self, name, power):
        super().__init__(name)
        self.__power = power
        self._bonus = 0

    def attack(self, enemy):
        damage = self.__power + self._bonus + randint(0, 15)
        enemy.health -= damage
        return f"{self.name} атакує мечем і наносить {damage} шкоди. HP {enemy.name}: {enemy.health}"

    def train(self):
        self._bonus += 2


# 🔹 Інший тип персонажа (поліморфізм)
class Barbarian(Character):
    def __init__(self, name, power):
        super().__init__(name)
        self.__power = power

    def attack(self, enemy):
        damage = self.__power + randint(0, 25)
        enemy.health -= damage
        return f"{self.name} атакує сокирою і наносить {damage} шкоди. HP {enemy.name}: {enemy.health}"


# 🔹 Створення персонажів
k = Knight("Рицар Артур", 100)
b = Barbarian("Варвар Кратос", 110)


# 🔹 Ігровий цикл
for turn in range(1, 11):
    print(f"\n🔸 Хід {turn}")

    k.train()
    print(k.attack(b))

    if b.health <= 0:
        print(f"🏆 Переміг {k.name}")
        break

    print(b.attack(k))

    if k.health <= 0:
        print(f"🏆 Переміг {b.name}")
        break
```

- Додамо третій тип персонажа — Archer (лучник), який:

- має власний параметр range_power (дальність атаки);
реалізує метод attack(), у якому шкода рахується за формулою
attack_power + randint(5, 15) + range_power;
має метод reload(), який збільшує дальність атаки (range_power += 1).

 - випадковий вибір персонажа з трьох варіантів: Knight, Barbarian, Archer;
покрокову гру, де користувач може виконувати дії під час свого ходу;
можливість вибору дії: атакувати або підсилити персонажа (наприклад, тренування або перезарядка), що впливає на наступні ходи;
ігровий цикл, у якому персонажі по черзі атакують один одного до перемоги одного з них.

```python
from abc import ABC, abstractmethod
from random import randint, choice


# 🔹 Абстрактний клас персонажа
class Character(ABC):
    def __init__(self, name, health=500):
        self.name = name
        self.health = health

    @abstractmethod
    def attack(self, enemy):
        pass

    def is_alive(self):
        return self.health > 0


# 🔹 Мечник
class Knight(Character):
    def __init__(self, name, power):
        super().__init__(name)
        self.__power = power
        self._bonus = 0

    def attack(self, enemy):
        damage = self.__power + self._bonus + randint(0, 10)
        enemy.health -= damage
        return f"{self.name} атакує мечем і завдає {damage} шкоди"

    def train(self):
        self._bonus += 2
        return f"{self.name} тренується (+2 до сили)"


# 🔹 Варвар
class Barbarian(Character):
    def __init__(self, name, power):
        super().__init__(name)
        self.__power = power

    def attack(self, enemy):
        damage = self.__power + randint(5, 20)
        enemy.health -= damage
        return f"{self.name} атакує сокирою і завдає {damage} шкоди"


# 🔹 Лучник (3 тип зброї)
class Archer(Character):
    def __init__(self, name, power, range_power=3):
        super().__init__(name)
        self.__power = power
        self.range_power = range_power

    def attack(self, enemy):
        damage = self.__power + randint(5, 15) + self.range_power
        enemy.health -= damage
        return f"{self.name} стріляє з лука і завдає {damage} шкоди"

    def reload(self):
        self.range_power += 1
        return f"{self.name} покращує дальність (+1)"


# 🔹 Створення персонажів
characters = [
    Knight("Рицар Артур", 100),
    Barbarian("Варвар Кратос", 110),
    Archer("Лучник Лін", 90)
]


# 🔹 Випадковий вибір двох бійців
p1, p2 = choice(characters), choice(characters)

while p1 == p2:
    p2 = choice(characters)

print(f"Бій: {p1.name} VS {p2.name}")


# 🔹 Ігровий цикл
turn = 1

while p1.is_alive() and p2.is_alive():
    print(f"\n🔸 Хід {turn}")

    # дія першого гравця
    if isinstance(p1, Knight):
        print(p1.train())
    elif isinstance(p1, Archer):
        print(p1.reload())

    print(p1.attack(p2))

    if not p2.is_alive():
        print(f"🏆 Переміг {p1.name}")
        break

    # дія другого гравця
    if isinstance(p2, Knight):
        print(p2.train())
    elif isinstance(p2, Archer):
        print(p2.reload())

    print(p2.attack(p1))

    if not p1.is_alive():
        print(f"🏆 Переміг {p2.name}")
        break

    turn += 1
```

```python
from random import choice, randint


# 🔹 Базовий клас зброї
class Weapon:
    def __init__(self, name, owner, health=500):
        self.name = name
        self.owner = owner
        self.health = health

    def attack(self, enemy):
        pass


# 🔹 Меч
class Sword(Weapon):
    def __init__(self, owner, power):
        super().__init__("Sword", owner)
        self.__power = power
        self._sharp = 0

    def attack(self, enemy):
        damage = self.__power + self._sharp + randint(0, 10)
        enemy.health -= damage
        return f"{self.owner} атакує мечем і завдає {damage} шкоди"

    def sharpening(self):
        self._sharp += 2
        return f"{self.owner} заточує меч (+2 сила)"


# 🔹 Сокира
class Axe(Weapon):
    def __init__(self, owner, power):
        super().__init__("Axe", owner)
        self.__power = power

    def attack(self, enemy):
        damage = self.__power + randint(5, 20)
        enemy.health -= damage
        return f"{self.owner} атакує сокирою і завдає {damage} шкоди"


# 🔹 Лук
class Bow(Weapon):
    def __init__(self, owner, power, range_power):
        super().__init__("Bow", owner)
        self.__power = power
        self.range_power = range_power

    def attack(self, enemy):
        damage = self.__power + randint(5, 15) + self.range_power
        enemy.health -= damage
        return f"{self.owner} стріляє з лука і завдає {damage} шкоди"

    def reload(self):
        self.range_power += 1
        return f"{self.owner} покращує дальність (+1)"


# 🔹 випадковий вибір зброї
def random_weapon(player_name):
    weapon_class = choice([Sword, Axe, Bow])

    if weapon_class == Sword:
        return Sword(player_name, 100)
    elif weapon_class == Axe:
        return Axe(player_name, 95)
    else:
        return Bow(player_name, 85, 3)


# 🔹 створення гравців
player = random_weapon("Гравець")
enemy = random_weapon("Бот")

print(f"Ваша зброя: {type(player).__name__}")
print(f"Зброя ворога: {type(enemy).__name__}")
print("=" * 40)

turn = 1

# 🔹 ігровий цикл
while player.health > 0 and enemy.health > 0:
    print(f"\nХід {turn}")
    print(f"Ваш HP: {player.health} | HP ворога: {enemy.health}")

    action = input("Оберіть дію (1-атака, 2-підсилення, 3-спец): ")

    if action == "1":
        print(player.attack(enemy))

    elif action == "2":
        if isinstance(player, Sword):
            print(player.sharpening())
        elif isinstance(player, Bow):
            print(player.reload())
        else:
            print("У цієї зброї немає підсилення")

    elif action == "3":
        print("Спеціальна дія недоступна в цій версії")

    else:
        print("Невірний вибір")

    if enemy.health <= 0:
        print("\n🎉 Ви перемогли!")
        break

    # 🔹 хід бота
    print(enemy.attack(player))

    if player.health <= 0:
        print("\n💀 Ви програли!")
        break

    turn += 1
```
